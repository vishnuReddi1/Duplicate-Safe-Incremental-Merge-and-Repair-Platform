```python
# Smart Data Deduplication Pipeline
# Notebook 1: Delta Setup

DATABASE_NAME = "smart_data_deduplication"

spark.sql(f"CREATE DATABASE IF NOT EXISTS {DATABASE_NAME}")

print("Database created/verified:", DATABASE_NAME)
```

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {DATABASE_NAME}.staging_records (
    record_id STRING,
    record_data STRING,
    source STRING,
    loaded_at TIMESTAMP
)
USING DELTA
""")

print("staging_records table ready!")
```

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {DATABASE_NAME}.clean_records (
    record_id STRING,
    record_data STRING,
    source STRING,
    updated_at TIMESTAMP
)
USING DELTA
""")

print("clean_records table ready!")
```

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {DATABASE_NAME}.duplicate_log (
    record_id STRING,
    duplicate_record_id STRING,
    similarity_score DOUBLE,
    duplicate_type STRING,
    root_cause STRING,
    action_taken STRING,
    detected_at TIMESTAMP
)
USING DELTA
""")

print("duplicate_log table ready!")
```

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {DATABASE_NAME}.quarantine_records (
    record_id STRING,
    record_data STRING,
    reason STRING,
    similarity_score DOUBLE,
    quarantined_at TIMESTAMP
)
USING DELTA
""")

print("quarantine_records table ready!")
```
